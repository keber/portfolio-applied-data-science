```python
# Función sigmoidea.

def sigmoidea(z):
    return 1 / (1 + np.exp(-z))
```

```python
# Función de loss

def loss(h, y):
    return (-y * np.log(h) - (1 - y) * np.log(1 - h)).mean()
```

```python
# Tu código aquí
def predict_probs(X, theta):
    return sigmoidea(np.dot(X, theta))

def add_intercept(X):
        intercept = np.ones((X.shape[0], 1))
        return np.concatenate((intercept, X), axis=1)

#gradiente = np.dot(x.T, (h - y)) / y.shape[0]
#lr = 0.01
#theta -= lr * gradiente

def fit(X, y, lr=0.01, epochs=20000):
    X = add_intercept(X)
    theta = np.zeros(X.shape[1])
    losses = []

    for _ in range(epochs):
        h = predict_probs(X, theta)
        gradiente = np.dot(X.T, (h - y)) / y.shape[0]
        theta -= lr * gradiente
        losses.append(loss(h,y))
    return theta, losses

# Entrenar el modelo manualmente
theta_manual, losses = fit(x_train, y_train)

# Graficar la función de pérdida
plt.plot(losses)
plt.xlabel("Iteraciones")
plt.ylabel("Pérdida")
plt.title("Función de pérdida en la regresión logística")
plt.show()

# Hacer predicciones en el conjunto de test
x_test_intercep = add_intercept(x_test)
preds_manual = predict_probs(x_test_intercep, theta_manual) >= 0.5

# Entrenar con sklearn
log_reg = LogisticRegression()
log_reg.fit(x_train, y_train)
preds_sklearn = log_reg.predict(x_test)

# Comparar resultados
print("Pesos con implementación manual:", theta_manual)
print("Pesos con sklearn:", log_reg.intercept_, log_reg.coef_)

# Calcular precisión de ambos modelos
accuracy_manual = np.mean(preds_manual == y_test)
accuracy_sklearn = np.mean(preds_sklearn == y_test)

print(f"Precisión implementación manual: {accuracy_manual:.4f}")
print(f"Precisión sklearn: {accuracy_sklearn:.4f}")
```
